````python
# Load data
import pandas as pd
data = pd.read_csv('voice.csv')
data.head()
````

### Langkah 3 - Preprocessing



````python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = data.iloc[:, :-1]   # semua kolom kecuali label
y = data.iloc[:, -1]    # kolom label terakhir

# Split data (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Standardisasi
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
````

### Langkah 4 - Buat Model kNN dan Evaluasi Awal

````python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Tentukan nilai K awal (misalnya 3)
knn = KNeighborsClassifier(n_neighbors=3)

# Latih model
knn.fit(X_train, y_train)

# Evaluasi model
y_pred = knn.predict(X_test)
print("Akurasi:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Laporan Klasifikasi:\n", classification_report(y_test, y_pred))
````

## 2. Percobaan Seleksi Fitur yang Paling Optimal



````python
# Encoding label ke biner (male: 1, female: 0) untuk analisis korelasi
data_corr = data.copy()
data_corr['label'] = data_corr['label'].map({'male': 1, 'female': 0})

# Hitung korelasi terhadap label dan urutkan
korelasi = data_corr.corr()['label'].abs().sort_values(ascending=False)
print("Korelasi Fitur terhadap Label:\n", korelasi)
````



### Pemilihan Fitur Terbaik

Berdasarkan pengujian beberapa kombinasi subset fitur, subset 6 fitur teratas memberikan akurasi tertinggi sekaligus membuat model lebih efisien:

- **meanfun** (mean fundamental frequency): korelasi tertinggi ($0.8339$), membedakan frekuensi nada dasar suara pria dan wanita secara signifikan.
- **IQR** (interquantile range): rentang persebaran frekuensi suara.
- **Q25** (first quantile): batas kuartil bawah frekuensi.
- **sp.ent** (spectral entropy): tingkat keacakan spektrum frekuensi suara.
- **sd** (standard deviation of frequency): simpangan baku frekuensi.
- **sfm** (spectral flatness): kerataan spektrum suara.

````python
# Fitur optimal yang dipilih
fitur_optimal = ['meanfun', 'IQR', 'Q25', 'sp.ent', 'sd', 'sfm']

X_opt = data[fitur_optimal]
y_opt = data['label']

# Split data menggunakan fitur optimal
X_train_opt, X_test_opt, y_train_opt, y_test_opt = train_test_split(
    X_opt, y_opt, test_size=0.3, random_state=42
)

# Standardisasi
scaler_opt = StandardScaler()
X_train_opt = scaler_opt.fit_transform(X_train_opt)
X_test_opt = scaler_opt.transform(X_test_opt)
````

## 3. Penentuan Nilai $k$ Terbaik



````python
import matplotlib.pyplot as plt

acc_test = []
acc_train = []

for k in range(1, 16):
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_opt, y_train_opt)
    acc_train.append(model.score(X_train_opt, y_train_opt))
    acc_test.append(model.score(X_test_opt, y_test_opt))

# Plot Grafik Analisis Nilai K
plt.figure(figsize=(8, 5))
plt.plot(range(1, 16), acc_train, marker='s', linestyle='--', label='Akurasi Train')
plt.plot(range(1, 16), acc_test, marker='o', label='Akurasi Test')
plt.title('Nilai K vs Akurasi (Fitur Optimal)')
plt.xlabel('Nilai K')
plt.ylabel('Akurasi')
plt.xticks(range(1, 16))
plt.legend()
plt.grid(True)
plt.show()
````

### Evaluasi Model Final dengan Nilai $k$ Terbaik ($k=8$)

````python
# Model final dengan k = 8
knn_optimal = KNeighborsClassifier(n_neighbors=8)
knn_optimal.fit(X_train_opt, y_train_opt)

y_pred_opt = knn_optimal.predict(X_test_opt)
print("Akurasi:", accuracy_score(y_test_opt, y_pred_opt))
print("Confusion Matrix:\n", confusion_matrix(y_test_opt, y_pred_opt))
print("Laporan Klasifikasi:\n", classification_report(y_test_opt, y_pred_opt))
````

## Kesimpulan dan Alasan Pemilihan Nilai $k=8$

- **Akurasi tertinggi:** Nilai $k = 8$ menghasilkan akurasi tertinggi pada data pengujian, yaitu $0.9832$ ($98.32\%$), meningkat dari akurasi awal baseline model ($97.69\%$).
- **Keseimbangan bias-variance:** Pada $k = 1$, model mengalami *overfitting* karena akurasi training mencapai 100% sementara akurasi testing lebih rendah. Pada $k = 8$, selisih performa antara training ($98.47\%$) dan testing ($98.32\%$) sangat kecil, menandakan model memiliki kemampuan generalisasi yang kuat tanpa *underfitting* maupun *overfitting*.
