```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv('wbc.csv')
```

### 1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.


```python
# Kolom 'id' tidak berguna untuk prediksi medis.
# Seringkali ada kolom 'Unnamed: 32' yang kosong karena koma berlebih di dataset.
cols_to_drop = ['id']
if 'Unnamed: 32' in df.columns:
    cols_to_drop.append('Unnamed: 32')

df = df.drop(columns=cols_to_drop)
```

### 2. Lakukan proses encoding pada kolom 'diagnosis'.



```python
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
print("\nLangkah 2 Selesai: Kolom 'diagnosis' telah di-encode.")
print("Mapping Label:", dict(zip(le.classes_, le.transform(le.classes_))))
```

### 3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

Kita akan menstandardisasi semua kolom fitur (selain 'diagnosis') agar memiliki rata-rata 0 dan standar deviasi 1. Ini penting untuk banyak algoritma Machine Learning.

```python
scaler = StandardScaler()

# Pisahkan fitur (X) dan target (y)
X = df.drop('diagnosis', axis=1)
y = df['diagnosis']

# Terapkan StandardScaler pada fitur
X_scaled = scaler.fit_transform(X)

# Konversi fitur yang diskalakan kembali ke DataFrame
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# Tampilkan beberapa baris pertama dari fitur yang diskalakan
print("Fitur yang Diskalakan (X_scaled_df):")
display(X_scaled_df.head())

print("Target (y):")
display(y.head())
```
